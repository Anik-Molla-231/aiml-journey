```python
from matplotlib import pyplot as plt
from matplotlib import style

style.use('ggplot')
x = [1,2,3]
y = [1,4,9]
x2 = [1,2,3]
y2 = [1,8,27]

plt.plot(x,y,label = 'square', linewidth =4, color = 'yellow')
plt.plot(x2,y2, label = 'cube')
plt.legend()
```
