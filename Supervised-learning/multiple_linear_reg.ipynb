```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error
```

# Importing the dataset

```python
df = pd.read_csv("Real-estate1.csv")

df.drop('No', inplace=True,axis=1)

print(df.head())
print(df.columns)
```

```python
# plotting a scatterplot
sns.scatterplot(x='X4 number of convenience stores',
                y='Y house price of unit area', data=df)
```

# Creating feature variables 

```python
X = df.drop('Y house price of unit area', axis=1)
y=df['Y house price of unit area']
print(X)
print(y)
```

```python
# creating train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=101)
```

```python
model = LinearRegression()

model.fit(X_train,y_train)

prediction = model.predict(X_test)
```

# Evaluation

```python
mae = mean_absolute_error(y_true=y_test, y_pred=prediction)
mse = mean_squared_error(y_true=y_test, y_pred=prediction)

print("Mean absolute error",mae)
print("Mean squared error", mse)
```
