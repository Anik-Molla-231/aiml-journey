```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

# 1. Handling Mising Data

```python
df = pd.read_csv("students.csv")
df["StudentID"].dropna()
df["Age"].fillna(df["Age"].mean(), inplace = True)
df["GPA"].fillna(0, inplace = True)
df.fillna("NaN", inplace = True)
```

# 2. Encoding

## (1) Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

df = pd.read_csv("students.csv")

df_label = df.copy()
le = LabelEncoder()

df_label["Gender Encoded"] = le.fit_transform(df["Gender"])

print("Label Encoded")
print(df_label[["Gender","Gender Encoded"]])
```

## (2) One Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

df = pd.DataFrame({
    'City': ['New York', 'Paris', 'Tokyo', 'Paris'],
    'Age': [25, 30, 22, 35]
})

df_encoded = pd.get_dummies(df, columns=['City'])
print("OneHot Encoded")
print(df_encoded)
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoder.set_output(transform='pandas')

encoded_city = encoder.fit_transform(df[['City']])
print(pd.concat([df.drop(columns='City'), encoded_city], axis=1))
```

# 3. Featrue Scaling

## Standard Scaling

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

scaler = StandardScaler().set_output(transform="pandas")
x,y = load_wine(return_X_y=True, as_frame=True)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.30, random_state=42, shuffle=False)

scaled_x_train = scaler.fit_transform(x_train)
```

## MinMaxScaling

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

scaler = MinMaxScaler().set_output(transform="pandas")
x,y = load_wine(return_X_y=True, as_frame=True)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.30, random_state=42, shuffle=False)

scaled_x_train = scaler.fit_transform(x_train)
```

## Robust Scaling

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import train_test_split

scaler = RobustScaler().set_output(transform="pandas")
x, y = load_wine(return_X_y=True, as_frame=True)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30,shuffle=False)

scaled_x = scaler.fit_transform(x_train)
```
