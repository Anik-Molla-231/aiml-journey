```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import statistics
```

# Arrays to be used

```python
# A basic dataset (e.g., student exam scores out of 100)
# Perfect for practicing basic center and spread (Mean, Median, Range)
scores = np.array([85, 92, 78, 88, 95, 89, 64, 78, 91, 83])

# A dataset with an outlier (e.g., annual salaries in thousands)
# Perfect for seeing how outliers skew the mean but leave the median unaffected
salaries = np.array([45, 50, 52, 48, 55, 47, 51, 320])

# A 2D array matrix (e.g., monthly sales for 3 different store branches)
# Perfect for practicing statistics along different axes (rows vs columns)
store_sales = np.array([
    [12, 15, 18, 14],  # Store A
    [10, 8,  12, 11],  # Store B
    [20, 22, 25, 21]   # Store C
])
```

# 1. Measures of Central Tendency

```python
mean = np.mean(scores)
print("Mean =", mean)
```

```python
median = np.median(scores)
print("Median =", median)
```

```python
mode = stats.mode(scores)
print("Mode =", mode)
```

```python
mean = np.mean(salaries)
print("Mean =", mean)
```

```python
median = np.median(salaries)
print("Median =", median)
```

# 2. Measure of Variability

```python
maximum = max(scores)
minimum = min(scores)
Range = maximum - minimum

print(f"Maximum = {maximum}, Minimum = {minimum}, Range = {Range}")
```

```python
var_scores = statistics.variance(scores)
var_salaries = statistics.variance(salaries)
print(f"Variance of scores: {var_scores} \nVariance of salaries: {var_salaries}")
```

```python
scores = [85, 92, 78, 88, 95, 89, 64, 78, 91, 83]
salaries = [45, 50, 52, 48, 55, 47, 51, 320]
std_scores = statistics.stdev(scores)
std_salaries = statistics.stdev(salaries)
print(f"Std_scores = {std_scores}, \nStd_salary = {std_salaries}")
```

## percentile

```python
data = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80]

# Calculate a single percentile (e.g., 75th percentile)
p75 = np.percentile(data, 75)
print(f"75th Percentile: {p75}")  # Output: 45.5

# Calculate multiple percentiles at once
p_multi = np.percentile(data, [25, 50, 75])
print(f"25th, 50th, 75th: {p_multi}")
```
