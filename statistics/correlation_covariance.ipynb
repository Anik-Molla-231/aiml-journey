```python
import pandas as pd
```

```python
hours = [1, 2, 3, 4, 5]
marks = [35, 45, 55, 65, 75]

df = pd.DataFrame({
    "hours": hours,
    "marks": marks})

print(df["hours"].cov(df["marks"]))
print(df["hours"].corr(df["marks"]))
```

```python
print(df.cov())
print(df.corr())
```

```python
x = [1, 2, 3, 4, 5]
y = [25, 16, 9, 4, 1]

df = pd.DataFrame({
    "x":x,
    "y":y)
print(df[x].corr(df[y]))
```
