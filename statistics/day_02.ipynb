# VARIENCE

```python
import numpy as np
import pandas as pd
```

```python
A = [48, 49, 50, 51, 52]
B = [10, 30, 50, 70, 90]
```

## varience

```python
var_A = np.var(A)
var_B = np.var(B)
print(var_A)
print(var_B)
```

## standard deviation

```python
std_A = np.std(A)
std_B = np.std(B)
print(std_A)
print(std_B)
```
